```python
import numpy as np
def ativacao(soma):
  if soma>=0:
      return 1
  else:
    return 0

def neuronio(x, w, theta):
  """
  Args:
    x: entradas
    W: vetor peso
    theta: vies
  Return:
  y: saida do neurônio
  """
  soma = np.dot(x,w) + theta
  y = ativacao(soma)
  return y
```

```python
x = np.array([1,0])
w = np.array([0.5,0.2])
theta = -0.5
saida = neuronio(x, w, theta)
print("a saída é:",saida)
```

```python
class Perceptron:
  def __init__ (self, input_size):
    self.weight = np.zeros(input_size) #w
    self.bias = 0

  def ativation(self, sum):
    if sum>=0:
      return 1
    else:
      return 0

  def output(self, inputs):
    sum = np.dot(inputs, self.weight) + self.bias
    output_perceptron = self.ativation(sum)
    return output_perceptron

  def train(self, input, output, epoch=10, learning_rate=1):
    for count in range(epoch):
      for input_in, output_in in zip(input, output): #automatiza o processo de iterar duas vezes
        output_perceptron = self.output(input_in)
        self.weight += learning_rate * (output_in - output_perceptron) * input_in #atualização dos pesos
        self.bias += learning_rate * (output_in - output_perceptron)
      print(count, " epoch === Erro:", output_in - output_perceptron)
```

```python
inputs = np.array([[0,0], [0,1], [1,0], [1,1]])
outputs = np.array([0,0,0,1])
perceptron = Perceptron(input_size=2)
perceptron.train(inputs, outputs, epoch=50, learning_rate=0.01)
```

```python
import matplotlib.pyplot as plt
def plot_hyperplane (weight, bias):
  input = np.linspace (-0.1, 1.1, 100)
  h = (-bias - weight[0]*input)/weight[1]#equação do hiperplano
  plt.plot(input, h, label = "Hiperplano encontrado", color = 'green')
```

```python
plot_hyperplane(perceptron.weight, perceptron.bias)
plt.style.use('ggplot')
plt.scatter(inputs[:,0], inputs[:,1], color = 'red')
plt.xlabel("Entrada 2")
plt.ylabel("Entrada 3")
```
